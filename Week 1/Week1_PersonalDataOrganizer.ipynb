## Task 4: Age-Related Calculations

Using Alex's fictional profile, this section calculates
his future age, past age, estimated days alive, and
years remaining until retirement.

```python
# Task 4: Age-Related Calculations

# Calculate future and past ages
future_age = age + 10
past_age = age - 5

# Estimate days alive, excluding leap years
days_alive = age * 365

# Calculate the remaining years until retirement
retirement_age = 65
years_to_retirement = retirement_age - age

# Display the calculated results
print("=== AGE CALCULATIONS ===")
print(f"Age in 10 years: {future_age}")
print(f"Age 5 years ago: {past_age}")
print(f"Estimated days alive: {days_alive:,}")
print(f"Years until retirement: {years_to_retirement}")
```

**Why do we need str(age) when combining an integer with a string using +, but not when using an f-string?**

It is because str(age) is used to explicitly cast the integer into a string so that Python is safely concatenating two identical data types. For example, when you use "Age: " + age, Python interprets the + operator as an instruction to either perform mathematical addition or string concatenation.
Because Python is strongly typed, it refuses to guess what you mean when you mix two completely different data types (a string and an integer).

Meanwhile, when you use f"Age: {age}", you are not using the addition operator at all. You are using string interpolation.
The entire purpose of an f-string is to construct a piece of text by evaluating the expressions inside the curly braces {} and embedding them

## Extra Challenge 3: Calculating Something Interesting

```python
#If Alex earns $50/hour, how much per day? Per year?

# Define Alex's hourly rate and working schedule
hourly_rate = 50
hours_per_day = 8
days_per_week = 5
weeks_per_year = 52

# Calculate daily, weekly, and annual earnings
daily_earnings = hourly_rate * hours_per_day
weekly_earnings = daily_earnings * days_per_week
annual_earnings = weekly_earnings * weeks_per_year

# Display the calculated earnings
print("=== ALEX'S EARNINGS CALCULATOR ===")
print(f"Hourly Rate: ${hourly_rate:.2f}")
print(f"Daily Earnings: ${daily_earnings:,.2f}")
print(f"Weekly Earnings: ${weekly_earnings:,.2f}")
print(f"Annual Earnings: ${annual_earnings:,.2f}")
```

```python
#If Alex drinks 2 liters of water/day, how much per year? 

# Define Alex's daily water consumption
water_per_day = 2
days_per_year = 365

# Calculate weekly, monthly, and annual consumption
weekly_water = water_per_day * 7
monthly_water = water_per_day * 30
annual_water = water_per_day * days_per_year

# Display the results
print("=== ALEX'S WATER CONSUMPTION ===")
print(f"Daily Consumption: {water_per_day} litres")
print(f"Weekly Consumption: {weekly_water} litres")
print(f"Monthly Consumption (30 days): {monthly_water} litres")
print(f"Annual Consumption: {annual_water:,} litres")
```
