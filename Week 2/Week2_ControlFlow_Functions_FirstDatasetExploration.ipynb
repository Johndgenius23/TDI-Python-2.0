## Task 1: Order Approval

```python
# Task 1: Order Approval Logic


def approval_message(order_value):
    if order_value > 100000:
        return "Needs manager approval"
    elif order_value > 50000:
        return "Needs supervisor approval"
    else:
        return "Auto-approved"

# Test the function
order_value = 72500

print(f"Order Value: {order_value:,}")
print(f"Approval Status: {approval_message(order_value)}")
```

## Task 2: Multiple Orders

```python
# Task 2: Process Multiple Orders

order_values = [12000, 55000, 150000, 49999, 100001, 8000]

for order in order_values:
    status = approval_message(order)
    print(f"Order Value: {order:,} | Status: {status}")
```

## Task 3: Supplier Status Function

```python
# Task 3: Supplier Status Function

def check_supplier_status(supplier_name, on_time_rate):
    if on_time_rate >= 0.90:
        status = "Reliable"
    elif on_time_rate >= 0.80:
        status = "Monitor"
    else:
        status = "At Risk"

    return f"{supplier_name}: {status}"

# Test with three suppliers

print(check_supplier_status("SupplierA", 0.94))
print(check_supplier_status("SupplierB", 0.86))
print(check_supplier_status("SupplierE", 0.74))
```

## Task 4: Load the Supply Chain Dataset

```python
# Task 4: Load the Supply Chain Dataset

import csv

with open("supply_chain_dataset.csv", newline="", encoding="utf-8-sig") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

# Display column names
print("=== DATASET COLUMNS ===")
print(list(rows[0].keys()))

# Display number of rows
print(f"\nTotal number of rows: {len(rows):,}")

# Display the first five rows
print("\n=== FIRST FIVE ROWS ===")

for row in rows[:5]:
    print(row)
```

## Task 5: Explore the Dataset

## Question 1: How many rows are there?


```python
# Question 1: Count the rows

total_rows = len(rows)

print(f"Total number of rows: {total_rows:,}")
```

## Question 2 : How many different supplier names appear?

```python
# Question 2: Find unique supplier names

supplier_names = []

for row in rows:
    supplier = row["supplier_name"]

    if supplier not in supplier_names:
        supplier_names.append(supplier)

print(f"Number of unique supplier names: {len(supplier_names)}")

print("\nSupplier Names:")
for supplier in supplier_names:
    print(supplier)
```

## Question 3: What is the highest total cost?

```python
# Question 3: Find the highest total cost

highest_cost = None

for row in rows:
    if row["total_cost"] != "":
        cost = float(row["total_cost"])

        if highest_cost is None or cost > highest_cost:
            highest_cost = cost

print(f"Highest Total Cost: {highest_cost:,.2f}")
```

## Question 4: What is the lowest total cost

```python
# Question 4: Find the lowest total cost

lowest_cost = None

for row in rows:
    if row["total_cost"] != "":
        cost = float(row["total_cost"])

        if lowest_cost is None or cost < lowest_cost:
            lowest_cost = cost

print(f"Lowest Total Cost: {lowest_cost:,.2f}")
```

## Question 5: How many orders were delivered?

```python
# Question 5: Count delivered orders

delivered_orders = 0

for row in rows:
    if row["delivery_status"] == "Delivered":
        delivered_orders += 1

print(f"Total Delivered Orders: {delivered_orders:,}")
```

## Extra Challenge 1: Advanced Order Approval

```python
# Extra Challenge 1: Advanced Order Approval

def advanced_approval(order_value, on_time_rate):

    if order_value > 100000 and on_time_rate < 0.80:
        return "Rejected: High value and unreliable supplier"

    elif order_value > 100000:
        return "Needs manager approval"

    elif order_value > 50000:
        return "Needs supervisor approval"

    else:
        return "Auto-approved"


# Test different scenarios
print(advanced_approval(150000, 0.75))
print(advanced_approval(150000, 0.95))
print(advanced_approval(75000, 0.85))
print(advanced_approval(25000, 0.70))
```

## Extra Challenge 2: Find All Orders From SupplierA

```python
# Extra Challenge 2: Find SupplierA Orders

supplier_a_count = 0
supplier_a_total_cost = 0

for row in rows:

    supplier_name = row["supplier_name"].lower().replace(" ", "")

    if supplier_name == "suppliera":
        supplier_a_count += 1

        if row["total_cost"] != "":
            supplier_a_total_cost += float(row["total_cost"])

print(f"Total SupplierA Orders: {supplier_a_count:,}")
print(f"Total SupplierA Order Cost: {supplier_a_total_cost:,.2f}")
```

## Extra Challenge 3: Find Missing Values

```python
# Extra Challenge 3: Identify Missing Values

missing_values = {}

for row in rows:
    for column in row:

        if row[column] == "":

            if column not in missing_values:
                missing_values[column] = 1

            else:
                missing_values[column] += 1

print("=== COLUMNS WITH MISSING VALUES ===")

for column, count in missing_values.items():
    print(f"{column}: {count:,} missing values")
```
